```python
import os
import pyspark
from pyspark.sql import SparkSession

# 1. Le decimos al sistema que apunte a la ruta dinámica donde pip instaló PySpark
print(os.path.dirname(pyspark.__file__))
print(pyspark.__path__[0])

```

```python
os.environ["SPARK_HOME"] = pyspark.__path__[0]

# 2. Ahora sí, levantamos la sesión normalmente
spark = SparkSession.builder \
    .appName("ReceptorMultiTabla") \
    .config("spark.driver.memory", "16g") \
    .config("spark.executor.memory", "16g") \
    .config("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension") \
    .config("spark.sql.catalog.spark_catalog", "org.apache.spark.sql.delta.catalog.DeltaCatalog") \
    .getOrCreate()

# 3. Y registramos las tablas para Metabase
spark.sql("""
    CREATE TABLE IF NOT EXISTS clientes 
    USING DELTA 
    LOCATION '/home/jovyan/data/processed/DataLake/clientes'
""")

spark.sql("""
    CREATE TABLE IF NOT EXISTS historiales 
    USING DELTA 
    LOCATION '/home/jovyan/data/processed/DataLake/historial'
""")

print("¡Tablas inscritas exitosamente en el catálogo de Spark!")
```

```python
spark.stop()
```
