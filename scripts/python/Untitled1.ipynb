```python
import os
import sys
import subprocess

# 1. Forzar limpieza de variables conflictivas de la imagen base
if "SPARK_HOME" in os.environ:
    del os.environ["SPARK_HOME"]

# 2. Capturar el comando exacto que intenta arrancar la JVM
from pyspark.java_gateway import launch_gateway
from pyspark import SparkConf

try:
    print("--- DETECTANDO ENTORNO ---")
    print(f"Versión de Python: {sys.version}")
    print(f"Ruta de Java (JAVA_HOME): {os.environ.get('JAVA_HOME')}")
    
    # Intentamos ejecutar Java directamente desde Python para ver su error real
    print("\n--- PRUEBA DIRECTA DE JAVA ---")
    resultado_java = subprocess.run(["java", "-version"], capture_output=True, text=True)
    print(resultado_java.stderr)
    
    print("\n--- PROBANDO ARRANQUE MANUAL DE LA JVM ---")
    conf = SparkConf()
    # Ejecutamos el gateway directo para capturar el desplome
    launch_gateway(conf)
    
except Exception as e:
    print("\n⚠️ LA JVM SE DESPLOMÓ POR LA SIGUIENTE RAZÓN:")
    print(str(e))
    
    # Comando desesperado pero efectivo: Leer los últimos logs del contenedor
    print("\n--- DIAGNÓSTICO DEL SISTEMA ---")
    if os.path.exists("/tmp"):
        logs = [f for f in os.listdir("/tmp") if "pyspark" in f]
        print(f"Archivos temporales de Spark encontrados: {logs}")
```
